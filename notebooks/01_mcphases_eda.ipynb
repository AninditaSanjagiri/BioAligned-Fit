```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
# 1. Load the new processed dataset
df = pd.read_csv('../data/processed/mcphases_cycle_syncing.csv')
```

```python
# 2. Target Distribution: Identifying Natural User Behavioral Imbalance

plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='exercise_target', palette='viridis')
plt.title('Real-World Distribution of Daily Exercise Intensity')
plt.xlabel('Intensity Level (0=None, 5=Very High)')
plt.show()
```

```python
# 3. Endocrinology Verification
# Proving Estrogen and Progesterone follow biological phase rules
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='phase', y='estrogen', ax=axes[0], palette='pastel')
axes[0].set_title('Estrogen (E3G) Distribution by Phase')

sns.boxplot(data=df, x='phase', y='pdg', ax=axes[1], palette='pastel')
axes[1].set_title('Progesterone (PdG) Distribution by Phase')
plt.tight_layout()
plt.show()
```

```python
# 4. Wearable Biometrics Verification
# Spotting the Luteal phase body temperature and resting heart rate spikes
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='phase', y='temperature_celsius', ax=axes[0], palette='coolwarm')
axes[0].set_title('Nightly Core Temperature by Phase')

sns.boxplot(data=df, x='phase', y='resting_heart_rate', ax=axes[1], palette='coolwarm')
axes[1].set_title('Resting Heart Rate by Phase')
plt.tight_layout()
plt.show()
```
